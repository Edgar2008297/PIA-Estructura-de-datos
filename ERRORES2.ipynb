```python
n = int(input('introduce un numero entero entre del 1 y 10: '))
file_name = 'tabla-' + str(n) + '.txt'
f = open(file_name, 'w')
for i in range(1, 11):
  f.write(str(n) + ' X ' + str(i) + ' = ' + str(n*i) + '\n')
f.close()
```
