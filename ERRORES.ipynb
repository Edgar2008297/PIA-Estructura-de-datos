```python
from logging import exception
import sys

lista_valores = ['x', 0, 4]

for valor in lista_valores:
  try:
    print(f"El valor en turno es {valor}")
    reciproco = 1 / int(valor)
    print(f"El valor del reciproco es {reciproco}")
  except excepcion:
    print(f"Ocurrio un problema {sys.exc_info()[0]}")
    excepcion = sys.exc_info()
    for elemento in excepcion:
      print(elemento)

  else:
        print("AHORA SI FUNCIONA")
```

```python
import sys


lista_valores = ['x', 0, 4]

for valor in lista_valores:
  try:
    print(f"El valor en turno es {valor}")
    reciproco = 1 / int(valor)
    print(f"El valor del reciproco es {reciproco}")
  except ValueError:
    print(f"El valor proporcionado ({valor}) no es compatible con la operacion solicitada")
  except ZeroDivisionError:
    print("la division sobre cero no es permitida")
  except Exception as e:
    print(f"Ocurrio un problema {sys.exc_info()[0]} {e}")
    excepcion = sys.exc_info()
    for elemento in excepcion:
      print(elemento)

  else:
        print("AHORA SI FUNCIONA")
```
