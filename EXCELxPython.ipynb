```python
import datetime
import openpyxl
import random
```

```python
libro = openpyxl.Workbook()

hoja = libro["Sheet"]
hoja.title = "Primera"
```

```python
hoja["B1"].value = "Listado de numeros"

for renglon in range (2,12):
  hoja.cell(row=renglon, column=2).value = random.randrange(100)
```

```python
hoja["c2"].value = "=SUM(B2:B11)"
```

```python
hoja["C15"].value = "Hoy es: "
hoja["D15"].value = datetime.datetime.now().strftime("%d/%m/%Y")
```
