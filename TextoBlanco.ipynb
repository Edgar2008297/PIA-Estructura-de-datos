```python
Entradas=[
    ['Correo', 'Nombre', 'Telefono'],
    ['juan@gmail.com', 'Juan', '8123232323'],
    ['maria@gmail.com', 'Maria', '5454545454'],
    ['diana@gmail.com', 'Diana', '4490909090']
]

for e in Entradas:
  print(f"{e[0]},{e[1]},{e[2]}")
```

```python
print(">>Contenido de la Lista.\n")
print(Entradas)

if os.path.exists("agenda.csv"):
  if(os.path.exists("agenda.bak")):
    os.remove("agenda.bak")
    os.rename("genda.csv", "agenda.bak")

with open("agenda.csv", "w+") as f:
  for e in Entradas:
    f.write(f"{e[0]},{e[1]},{e[2]}\n")
```

```python
datos = []
with open("agenda.csv", "r") as f:
  for linea in f:

    lista=linea.split(",")

    lista[2]=lista[2].replace("\n", "")
    datos.append(lista)

print(">>Contenido a la nueva lista.\n")
print(datos)
```
