```python
import openpyxl
```

```python
libro = openpyxl.Workbook()
hoja = libro.active
hoja.title = "Alumnos"
print("Archivo de Excel creado")
```

```python
hoja["A1"] = "Nombre"
hoja["B1"] = "Edad"
hoja["C1"] = "Carrera"
hoja["A2"] = "Ana"
hoja["B2"] = 20
hoja["C2"] = "Administracion"
hoja["A3"] = ""
```

```python
libro.save("alumnos.xlsx")
print("Archivo guradado correctamente")
```

```python
libro.close()
print("Archivo Cerrado")
```

```python
libro = openpyxl.load_workbook("alumnos.xlsx")
hoja = libro.active
print("Archivo de Excel abierto")


```

```python
for fila in hoja.iter_rows(values_only=True):
  print(fila)
```
