```python
import pandas as pd

notas = pd.Series([87, 100, 85, 60, 78])
print(type(notas))
print(notas, "\n")

iguales = pd.Series(100, range (6))
print(type(iguales))
print(iguales)
```

```python
print("notas : ")
print(f"{notas}")

print(f"\ncantidad = {notas.count()}")
print(f"media = {notas.mean()}")
print(f"mediana = {notas.median()}")
print(f"mínimo = {notas.min()}")
print(f"maximo = {notas.max()}")
print(f"desviación estándar = {notas.std():.6f}", "\n")

print("\nSumariacion descriptiva:")
print(notas.describe())
print(type(notas.describe))
```

```python
import pandas as pd
SEPARADOR = ("*" * 60) + "\n"


diccionario_notas = {"Programacion":[87, 80, 80, 20, 100],
                     "Base de Datos": [71, 100, 70, 100, 100],
                     "Contabilidad":[99, 57, 57, 100, 100]}

notas_diccionario = pd.DataFrame(diccionario_notas, index=["Crescencio", "Domitila", "Rutilio", "Panfilo", "Ludoviko"])
print(notas_diccionario)
print(SEPARADOR)
```

```python
print("\nLas notas obtenidas en programacion fueron:")

print(notas_diccionario["Programacion"])
print()

print(notas_diccionario.Programacion)
print(SEPARADOR)
```

```python
print("las notas obtenidas por Rutilio fueron:")
print(notas_diccionario.loc["Rutilio"])
print("\nLas notas obtenidas por Ludoviko fueron:")
print(notas_diccionario.loc["Ludoviko"])
```

```python
print(f"La menor calificacion de Crescencio fue de {notas_diccionario.loc['Crescencio'].min()}")
print()
print(f"La mayor calificacion de Crescencio fue de {notas_diccionario.loc['Crescencio'].mean():.4f}")
print()
```

```python
medias = notas_diccionario.mean()
print(type(medias))
print("\nPromedios por UA")
print("Se obtuvo la media de las siguientes asignaturas:")

for materia in medias.index:
  print(materia)

print(f"\nEl promedio de las notas obtenidas fue de")
print(medias)
```

```python
media_global = notas_diccionario.mean().mean()
print(f"\nLa media global del DataFrame es de {media_global:.4f}")
```

```python
print("las primeras dos materias para todos los estudiantes")
print(notas_diccionario[["Programacion", "Base de Datos"]])

print("\nLos ultimos dos estudiantes")
print(notas_diccionario.loc["Panfilo":"Ludoviko"])

print("\nLos estudiantes hasta Rutilio")
print(notas_diccionario.loc[:"Rutilio"])
```

```python
print("\nUltimas dos asignaturas para todos los estudiantes a partir de Rutilio")
print(notas_diccionario[["Base de Datos", "Contabilidad"]].loc["Rutilio":])

print(type(notas_diccionario[["Base de Datos", "Contabilidad"]].loc["Rutilio":]))
print(SEPARADOR)
```

```python
print("La calificacion de Panfilo en Programacion fue de:")
print(notas_diccionario.at["Panfilo", "Programacion"])
```

```python
filtro = notas_diccionario > 79

print("Este seria el filtro o mapa de bits resultante de aplicar la condicion")
print(filtro)
```

```python
print("Este seria el resultado de aplicar el filtro o mapa de bits sobre el DataFrame")
print(notas_diccionario[notas_diccionario < 70])
print(SEPARADOR)

print(notas_diccionario[notas_diccionario < 70].min())
```

```python
base_datos_70_71 = notas_diccionario.loc[notas_diccionario["Base de Datos"].isin([70,71])]

print(base_datos_70_71)
```

```python
filtro_aprobados_menor_100 = (notas_diccionario["Programacion"] >= 70) & (notas_diccionario["Programacion"] < 100)
aprobaron_progra_menos_100 = notas_diccionario.loc[filtro_aprobados_menor_100]

print(aprobaron_progra_menos_100)
print(SEPARADOR)
print(aprobaron_progra_menos_100["Programacion"])
```
