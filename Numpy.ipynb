```python
datos = np.random.randint(1,200, size=(18,))
print("los datos son: ")
print(datos)

minimo = datos.min()
maximo = datos.max()

print(f"\nLos datos tienen un rango de {maximo - minimo}, abarcando desde el {minimo} hasta el {maximo}")
```

```python
#Por metodo Libreria
minimo = np.min(datos)
maximo = np.max(datos)

print(f"\nLos datos tienen un rango de {maximo - minimo}, abarcando desde el {minimo} hasta el {maximo}")

primer_cuartil = np.percentile(datos, 25)
mediana = np.median(datos)
tercer_cuartil = np.percentile(datos, 75)

print(f"\nLa mediana de datos es igual a {mediana}")
print(f"El primer cuartil de datos es igual a {primer_cuartil}, mientras que el tercer cuartil es igual a {tercer_cuartil}")
print(f"Con rango inercuartil = {primer_cuartil} a {tercer_cuartil} o {tercer_cuartil - primer_cuartil}")
```

```python
#por metodo directo de arreglo
media = datos.mean()
desviacion_estandar = datos.std()
varianza = datos.var()


print(f"\nLos datos poseen una media aritmetica de {media:.4f}, con una varianza de {varianza:.4f} y una desviacion estandar de {desviacion_estandar:.4f}")
```

```python
#por metodo libreria
mediana = np.mean(datos)
desviacion_estandar = np.std(datos)
varianza = np.var(datos)
print(f"\nLos datos poseen una media aritmetica de {media:.4f}, con una varianza de {varianza:.4f} y una desviacion estandar de {desviacion_estandar:.4f}")
```

```python
import numpy as np

notas = np.array([78,92, 100, 88, 70])
promedio = np.mean(notas)
max = np.max(notas)
min = np.min(notas)
cantidad = np.sum(notas >= 70)

print("promedio: ", promedio)
print("maximo: ", max)
print("minimo: ", min)
print("notas >= 70: ", cantidad)

if promedio >= 70:
  print("El promedio alcanza 70")
else:
  print("El promedio no alcanza a 70 XD")
```

```python
import numpy as np

temperaturas = np.array([18, 21, 25, 19, 23, 27, 20])
promedio = np.mean(temperaturas)
dias_encima = np.sum(temperaturas > promedio)
fahrenheit = temperaturas * 9/5 + 32

print("promedio: ", promedio)
print("Dias por encima: ", dias_encima)
print("Fahrenheit: ", fahrenheit)
```
