```python
import csv
import sys
datos_a_leer = dict()

try:
  with open("autos.csv","r", encoding='latin1',newline="") as archivo:
    lector = csv.reader(archivo)
    next(lector)

    for clave, marca, año in lector:
      datos_a_leer[int(clave)] = (marca, int(año))
except FileNotFoundError:
  print("El archivo no se encontro, se procede a trabajar en conjunto vacio")
except Exception:
  Excepcion = sys.exc_info()
  print(f"Ocurrio un error del tipo: {Excepcion[0]}")
  print(f"Con el mensaje: {Excepcion[1]}")

else:
  print(datos_a_leer)
```

```python
import sys
datos_a_leer = dict()

try:
  with open("autos.csv","r", encoding='latin1',newline="") as archivo:
    lector = csv.reader(archivo)
    next(lector)

    for clave, marca, año in lector:
      datos_a_leer[int(clave)] = (marca, int(año))
except FileNotFoundError:
  print("El archivo no se encontro, se procede a trabajar en conjunto vacio")
except csv.Error as fallo_csv:
  print(f"Ocurrio un error de tipo csv: {fallo_csv}")
except Exception:
  Excepcion = sys.exc_info()
  print(f"Ocurrio un error del tipo: {Excepcion[0]}")
  print(f"Con el mensaje: {Excepcion[1]}")

else:
  print(datos_a_leer)
```
