```python
lista_base = ["a", "b", "c", "d"]
copia = lista_base

copia_2 = lista_base.copy()
copia_3 = list(lista_base)
copia_4 = lista_base[:]
print(f"lista base = {lista_base}, y la copia es {copia}")
print(f"lista base = {lista_base}, y la copia 2 es {copia_2}")
print(f"lista base = {lista_base}, y la copia 3 es {copia_3}")
print(f"lista base = {lista_base}, y la copia 4 es {copia_4}")
```

```python
Lista_uno = [10, 20, 30, 40, 50]
lista_resultante = []
for valor in Lista_uno:
  lista_resultante.append(valor * 2)
print(f"Resultado = {lista_resultante}")
```

```python
Lista_uno = [10, 20, 30, 40, 50]
lista_con_resultado = [valor * 2 for valor in Lista_uno]
print(f"Resultado = {lista_con_resultado}")
```
